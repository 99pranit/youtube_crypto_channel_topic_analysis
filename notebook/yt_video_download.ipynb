# Import Libraries

```python
from pytube import YouTube
import yt_dlp
import time
import pandas as pd
import os
import shutil
```

```python
filepath = os.getcwd().replace('youtube_crypto_channel_topic_analysis/notebook' , '') + '3'
delay = 2
```

```python
ydl_opts = {
    'cookiefile': 'cookies.txt',
    'format': 'bestvideo[height<=360]+bestaudio/best[height<=360]',
    'outtmpl': os.path.join(filepath, '%(id)s.%(ext)s'),
    'retries': 5,
    'fragment_retries': 5,
    'ignoreerrors': True,
    'no_warnings': True,
    'quiet': True,
    'sleep_interval': delay
}
```

# Modules

```python
def download_videos(video_ids):
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        for i, video_id in enumerate(video_ids):
            url = f'https://www.youtube.com/watch?v={video_id}'
            print(f'Downloading {i+1}/{len(video_ids)}: {video_id}')
            
            try:
                ydl.download([url])
                print(f"✅ Downloaded successfully")
            except Exception as e:
                print(f"🔥 Failed: {str(e)}")
            print('-' * 50)
```

# download videos

```python
yt_df = pd.read_csv('LDA_yt_list.csv')

video_ids = yt_df['id'][450:500]

download_videos(video_ids)
print("All downloads completed!")
```
